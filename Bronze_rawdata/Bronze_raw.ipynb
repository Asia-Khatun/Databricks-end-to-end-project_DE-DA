# Bronze Layer - Raw Data Ingestion

This notebook reads raw CSV files from Unity Catalog volumes and loads them into the Bronze layer tables in the `end-to-end_pipeline.bronze` schema.

**Tables ingested:**
* Sales Transactions (25,005 rows)
* Stores (30 rows)
* Suppliers (41 rows)
* Products (251 rows)
* Customers (2,501 rows)
* Calendar (1,826 rows)

# Sales Transactions Table
Reading sales transactions data from CSV and loading to Bronze layer

```python
df = (spark.read.option("header", "true").option("inferSchema", "true")
  .csv("/Volumes/end-to-end_pipeline/bronze/retail_source_rawdata/sales_transactions_raw.csv")
)
display(df)
```

## Write to Bronze Layer

```python
df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("`end-to-end_pipeline`.bronze.sales_transactions")
```
