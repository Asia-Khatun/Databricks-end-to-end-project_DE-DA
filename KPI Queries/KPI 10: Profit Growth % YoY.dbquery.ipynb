```sql
-- KPI 10: Profit Growth % YoY
-- Purpose:
-- Calculates year-over-year profit growth by comparing
-- the latest year's profit with the previous year's profit.

WITH yearly_profit AS (
    SELECT
        YEAR(order_date) AS sales_year,
        SUM(profit) AS total_profit
    FROM `end-to-end_pipeline`.gold.fact_sales
    GROUP BY YEAR(order_date)
),

profit_comparison AS (
    SELECT
        sales_year,
        total_profit,
        LAG(total_profit) OVER (ORDER BY sales_year) AS previous_year_profit
    FROM yearly_profit
)

SELECT
    sales_year,
    ROUND(
        ((total_profit - previous_year_profit)
        / NULLIF(previous_year_profit, 0)) * 100,
        2
    ) AS profit_growth_yoy_pct
FROM profit_comparison
WHERE previous_year_profit IS NOT NULL
ORDER BY sales_year DESC
LIMIT 1;
```
