```sql
-- Analysis: Return & Cancellation by Region
-- Purpose:
-- Measures return and cancellation rates across regions
-- to identify potential operational performance issues.

SELECT
    s.region,

    COUNT(DISTINCT f.order_id) AS total_orders,

    COUNT(
        DISTINCT CASE
            WHEN f.order_status = 'Returned'
            THEN f.order_id
        END
    ) AS returned_orders,

    COUNT(
        DISTINCT CASE
            WHEN f.order_status = 'Cancelled'
            THEN f.order_id
        END
    ) AS cancelled_orders,

    ROUND(
        COUNT(
            DISTINCT CASE
                WHEN f.order_status = 'Returned'
                THEN f.order_id
            END
        ) * 100.0
        / NULLIF(COUNT(DISTINCT f.order_id), 0),
        2
    ) AS return_rate_pct,

    ROUND(
        COUNT(
            DISTINCT CASE
                WHEN f.order_status = 'Cancelled'
                THEN f.order_id
            END
        ) * 100.0
        / NULLIF(COUNT(DISTINCT f.order_id), 0),
        2
    ) AS cancellation_rate_pct

FROM `end-to-end_pipeline`.gold.fact_sales f

INNER JOIN `end-to-end_pipeline`.gold.dim_store s
    ON f.store_id = s.store_id

GROUP BY s.region

ORDER BY return_rate_pct DESC;
```
