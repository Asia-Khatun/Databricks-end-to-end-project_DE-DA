```sql
-- Analysis: Regional Profitability
-- Purpose:
-- Compares revenue, profit, and profit margin across regions
-- to identify high-value and low-margin regional markets.

SELECT
    s.region,

    ROUND(SUM(f.net_revenue), 2) AS total_revenue,

    ROUND(SUM(f.profit), 2) AS total_profit,

    ROUND(
        (SUM(f.profit) / NULLIF(SUM(f.net_revenue), 0)) * 100,
        2
    ) AS profit_margin_pct,

    COUNT(
        DISTINCT CASE
            WHEN f.order_status <> 'Cancelled'
            THEN f.order_id
        END
    ) AS total_orders

FROM `end-to-end_pipeline`.gold.fact_sales f

INNER JOIN `end-to-end_pipeline`.gold.dim_store s
    ON f.store_id = s.store_id

GROUP BY s.region

ORDER BY total_revenue DESC;
```
