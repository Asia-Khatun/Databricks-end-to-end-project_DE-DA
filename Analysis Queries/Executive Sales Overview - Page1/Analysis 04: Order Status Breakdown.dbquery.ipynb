```sql
-- Analysis 04: Order Status Breakdown
-- Purpose:
-- Analyzes the number and percentage of orders
-- by Completed, Cancelled, and Returned status.

WITH order_status_summary AS (
    SELECT
        order_status,
        COUNT(DISTINCT order_id) AS total_orders
    FROM `end-to-end_pipeline`.gold.fact_sales
    GROUP BY order_status
)

SELECT
    order_status,
    total_orders,
    ROUND(
        total_orders * 100.0 / SUM(total_orders) OVER (),
        2
    ) AS order_percentage
FROM order_status_summary
ORDER BY total_orders DESC;
```
