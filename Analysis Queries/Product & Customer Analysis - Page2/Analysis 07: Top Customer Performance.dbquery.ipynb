```sql
-- Analysis 07: Top Customer Performance
-- Purpose:
-- Identifies the top 10 customers by revenue and shows
-- their revenue, profit, and order contribution.

WITH customer_performance AS (
    SELECT
        c.customer_id,
        c.customer_name,
        c.customer_segment,

        ROUND(SUM(f.net_revenue), 2) AS total_revenue,

        ROUND(SUM(f.profit), 2) AS total_profit,

        COUNT(DISTINCT CASE
            WHEN f.order_status <> 'Cancelled'
            THEN f.order_id
        END) AS total_orders

    FROM `end-to-end_pipeline`.gold.fact_sales f

    INNER JOIN `end-to-end_pipeline`.gold.dim_customer c
        ON f.customer_id = c.customer_id

    GROUP BY
        c.customer_id,
        c.customer_name,
        c.customer_segment
)

SELECT
    customer_id,
    customer_name,
    customer_segment,
    total_revenue,
    total_profit,
    total_orders

FROM customer_performance

ORDER BY total_revenue DESC

LIMIT 10;
```
