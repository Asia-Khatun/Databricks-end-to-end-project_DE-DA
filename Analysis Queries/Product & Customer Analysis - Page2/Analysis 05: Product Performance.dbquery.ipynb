```sql
-- Analysis 05: Product Performance
-- Purpose:
-- Analyzes the top 10 products by revenue, including
-- revenue, cost, profit, and profit margin percentage.

WITH product_performance AS (
    SELECT
        p.product_id,
        p.product_name,
        p.category,

        ROUND(SUM(f.net_revenue), 2) AS total_revenue,

        ROUND(SUM(f.total_cost), 2) AS total_cost,

        ROUND(SUM(f.profit), 2) AS total_profit,

        ROUND(
            (SUM(f.profit) / NULLIF(SUM(f.net_revenue), 0)) * 100,
            2
        ) AS profit_margin_pct

    FROM `end-to-end_pipeline`.gold.fact_sales f

    INNER JOIN `end-to-end_pipeline`.gold.dim_product p
        ON f.product_id = p.product_id

    GROUP BY
        p.product_id,
        p.product_name,
        p.category
)

SELECT
    product_id,
    product_name,
    category,
    total_revenue,
    total_cost,
    total_profit,
    profit_margin_pct

FROM product_performance

ORDER BY total_revenue DESC

LIMIT 10;
```
