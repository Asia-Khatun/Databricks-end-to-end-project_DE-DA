```sql
-- Analysis 02: Category Performance
-- Purpose:
-- Analyzes revenue, cost, profit, margin, orders,
-- and quantity sold by product category.

SELECT
    p.category,

    ROUND(SUM(f.net_revenue), 2) AS total_revenue,

    ROUND(SUM(f.total_cost), 2) AS total_cost,

    ROUND(SUM(f.profit), 2) AS total_profit,

    ROUND(
        (SUM(f.profit) / NULLIF(SUM(f.net_revenue), 0)) * 100,
        2
    ) AS profit_margin_pct,

    COUNT(DISTINCT CASE
        WHEN f.order_status <> 'Cancelled'
        THEN f.order_id
    END) AS total_orders,

    SUM(
        CASE
            WHEN f.order_status = 'Completed'
            THEN f.quantity
            ELSE 0
        END
    ) AS units_sold

FROM `end-to-end_pipeline`.gold.fact_sales f

INNER JOIN `end-to-end_pipeline`.gold.dim_product p
    ON f.product_id = p.product_id

GROUP BY p.category

ORDER BY total_revenue DESC;
```
