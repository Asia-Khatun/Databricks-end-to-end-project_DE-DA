```sql
-- Analysis 06: Customer Segment Performance
-- Purpose:
-- Analyzes revenue, cost, profit, profit margin,
-- customers, and orders by customer segment.

SELECT
    c.customer_segment,

    COUNT(DISTINCT f.customer_id) AS total_customers,

    COUNT(DISTINCT CASE
        WHEN f.order_status <> 'Cancelled'
        THEN f.order_id
    END) AS total_orders,

    ROUND(SUM(f.net_revenue), 2) AS total_revenue,

    ROUND(SUM(f.total_cost), 2) AS total_cost,

    ROUND(SUM(f.profit), 2) AS total_profit,

    ROUND(
        (SUM(f.profit) / NULLIF(SUM(f.net_revenue), 0)) * 100,
        2
    ) AS profit_margin_pct

FROM `end-to-end_pipeline`.gold.fact_sales f

INNER JOIN `end-to-end_pipeline`.gold.dim_customer c
    ON f.customer_id = c.customer_id

GROUP BY
    c.customer_segment

ORDER BY
    total_revenue DESC;
```
